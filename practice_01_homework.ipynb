### Задание 2 (1)

Извлеките из текста даты в годах, не извлекая другие числа. Извлекаться должен только сам год (в случае 1920-x извлекать -x не нужно).
В случае 2000-2010 нужно извлечь два года. 

```python
TEXT_WIKI = """Первым открытым белым карликом[3] стала звезда 40 Эридана B, которую ещё в 1785 году
Уильям Гершель включил в каталог двойных звёзд[4]. В 1910 году Генри Норрис Расселл
обратил внимание на аномально низкую светимость 40 Эридана B (при её высокой цветовой
температуре), что и послужило выделению таких звёзд в отдельный класс.

В январе 1862 года Элвин Грэхэм Кларк, юстируя 18-дюймовый рефрактор (самый большой на
то время телескоп в мире), обнаружил в непосредственной близости от Сириуса тусклую
звёздочку[6]. А в 1896 году американский астроном Д. М. Шеберле открыл Процион B,
подтвердив тем самым второе предсказание Бесселя.

В 1915 году Уолтер Сидней Адамс измерил спектр Сириуса B: температура поверхности
Сириуса B составляет 25 000 K, а Сириуса A — 10 000 К, плотность — 106 г/см3
(плотность Солнца ~1,4 г/см3). В 1922 году Виллем Якоб Лейтен предложил называть такие
звёзды "белыми карликами"[8]. Стандартная модель строения звёзд была разработана в
1920-х годах Эддингтоном, а в 1926 году Фаулер показал, что для белых карликов
давление определяется свойствами вырожденного электронного газа[10].

В 1999—2000 годах галилеевы спутники наблюдала обсерватория "Чандра"[121]. Аппарат
"Юнона", запущенный 5 августа 2011 года, 29 сентября 2022 года пролетел на расстоянии
352 километров от поверхности Европы. 14 апреля 2023 года состоялся запуск аппарата
Jupiter Icy Moons Explorer (JUICE); прибытие в систему Юпитера ожидается в 2031 году,
и уже через год состоятся два пролёта Европы на расстоянии 400—500 километров.

Поверхность спутника нагрета до 1420 К, в каталоге числится 21922 объекта,
а подробный обзор занимает стр. 1920—1925 сборника."""
```

```python
# 4-значные числа, перед которыми стоит пробел и после — "год"/"году"/"годах"
# или диапазоны вида 1999—2000, 1920—1925
pattern = r'(1[0-9]{3}|20[0-9]{2})(?=\s*(?:год|году|годах|—|-))'
years = re.findall(pattern, TEXT_WIKI)

# Для диапазонов нужно также извлечь второй год
pattern_range = r'(1[0-9]{3}|20[0-9]{2})—((?:1[0-9]{3}|20[0-9]{2}))(?=\s*(?:годах|—|-))'
ranges = re.findall(pattern_range, TEXT_WIKI)

# Итоговый список всех годов
all_years = []
for y in years:
    all_years.append(y)
for a, b in ranges:
    all_years.append(a)
    all_years.append(b)

# Убираем дубликаты, сохраняя порядок
seen = set()
result = []
for y in all_years:
    if y not in seen:
        seen.add(y)
        result.append(y)

print(result)
```

## Задание 5 (1) 

Напишите такую регулярку, которая извлечет ФИО в каноническом виде из примеров ниже.   
"Иванов И. И." -> должно нормализоваться до "Иванов И.И." (Фамилия + Инициалы без пробелов)  
"И. И. Иванов" - "Иванов И.И." 

Подсказка: напишите отдельные регулярки для инициалов и фамилий, затем напишите еще одну регулярку, которая включает возможные комбинации и порядки. Используйте именованные группы, чтобы преобразовать части в каноничный порядок. 
В коде не должно быть никаких дополнительных нормализаций (дополнительный re.sub или .replace и т.п)

```python
NAMES = [
    "Иванов И. И.",
    "И. И. Иванов",
    "Иванов И.И.",
    "А.С. Пушкин",
    "Пушкин А.С.",
    "Д. М. Шеберле",
    "М. Кюри",
    "Ж.К. Ван Дамм",
]
```

```python
# Инициал: заглавная буква + точка
INITIAL = r'[А-ЯЁ]\.'

# Фамилия: одно или несколько слов
SURNAME = r'(?:[А-ЯЁ][а-яё]+)(?:\s+[А-ЯЁ][а-яё]+)*'

# Вариант 1: Фамилия + инициалы  → "Иванов И. И." / "Иванов И.И."
# Вариант 2: Инициалы + Фамилия  → "И. И. Иванов" / "И.И. Иванов"

pattern = re.compile(rf"""
    (?:
        # Вариант 1: Фамилия впереди
        (?P<surname1>{SURNAME})
        \s+
        (?P<init1>{INITIAL})
        \s*
        (?P<init2>{INITIAL})?
    )
    |
    (?:
        # Вариант 2: Инициалы впереди
        (?P<init3>{INITIAL})
        \s*
        (?P<init4>{INITIAL})?
        \s+
        (?P<surname2>{SURNAME})
    )
""", re.VERBOSE)

def normalize(match):
    # Определяем, какой вариант сработал
    if match.group('surname1'):
        surname = match.group('surname1')
        inits = [match.group('init1'), match.group('init2')]
    else:
        surname = match.group('surname2')
        inits = [match.group('init3'), match.group('init4')]

    # Собираем инициалы без пробелов: "И.И."
    initials = ''.join(i for i in inits if i)
    return f"{surname} {initials}"

result = [pattern.sub(normalize, name) for name in NAMES]

for src, dst in zip(NAMES, result):
    print(f"{src} -> {dst}")
```
